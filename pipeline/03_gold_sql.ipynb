## Gold transformations

### 1) Basic rollups

```sql
USE CATALOG workouts_demo;
USE SCHEMA gold;

CREATE OR REPLACE VIEW workouts_demo.gold.summary_by_sport AS
SELECT
  sport,
  COUNT(*)                                   AS activities,
  ROUND(AVG(avg_speed_mps * 3.6), 2)         AS avg_speed_kmh,
  ROUND(AVG(distance_m / 1000), 2)           AS avg_distance_km,
  ROUND(AVG(moving_time_s / 3600), 2)        AS avg_moving_h,
  ROUND(AVG(elapsed_time_s / 3600), 2)       AS avg_elapsed_h,
  ROUND(AVG(elevation_m), 2)                    AS avg_elevation_m,
  AVG(CASE WHEN elapsed_time_s > 0 THEN ROUND(moving_time_s * 1.0 / elapsed_time_s, 3) END)
                                             AS avg_moving_ratio  -- “how much of elapsed was moving”
FROM workouts_demo.silver.activities
GROUP BY sport
ORDER BY activities DESC;
```

### 2) Longest efforts

Could be used for queries about long workouts

```sql
CREATE OR REPLACE VIEW workouts_demo.gold.longest_activities AS
SELECT
  activity_id,
  sport,
  moving_time_s,
  elapsed_time_s,
  ROUND(distance_m / 1000, 3) AS distance_km,
  ROUND(moving_time_s / 3600, 3) AS moving_time_h
FROM workouts_demo.silver.activities
ORDER BY moving_time_s DESC
LIMIT 100; 
```
