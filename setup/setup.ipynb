## SQL setup notebook
- Creates the catalog
- Creates schemas (`bronze`, `silver`, `gold`)
- Creates the landing volume into `bronze`
- Grants privileges

```sql
DROP CATALOG IF EXISTS workouts_demo CASCADE;
```

```sql
CREATE CATALOG workouts_demo COMMENT "Demo catalog of workout data";
```

```sql
CREATE SCHEMA workouts_demo.bronze COMMENT "Bronze schema for raw data";
CREATE SCHEMA workouts_demo.silver COMMENT "Silver schema for validated data";
CREATE SCHEMA workouts_demo.gold COMMENT "Gold schema";
```

```sql
CREATE VOLUME workouts_demo.bronze.landing COMMENT "Landing volume (raw files)";
```

```sql
GRANT ALL PRIVILEGES ON CATALOG workouts_demo TO `juhotimonen50@gmail.com`;
```
