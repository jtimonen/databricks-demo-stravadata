## SQL setup notebook
- Creates the catalog
- Creates schemas (`bronze`, `silver`, `gold`)
- Creates the landing volume into `bronze`
- Creates empty Delta tables
- Grants privileges

```sql
-- Catalog & schemas
CREATE CATALOG IF NOT EXISTS workouts_demo;
USE CATALOG workouts_demo;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;

-- Volume to drop CSVs into
CREATE VOLUME IF NOT EXISTS bronze.landing
COMMENT 'CSV drop zone: /Volumes/workouts_demo/bronze/landing';

-- Empty Delta tables
CREATE TABLE IF NOT EXISTS workouts_demo.bronze.activities_raw
USING DELTA
COMMENT 'Raw Strava activities loaded via COPY INTO';

CREATE TABLE IF NOT EXISTS workouts_demo.silver.activities
USING DELTA
COMMENT 'Cleaned, typed, deduped activities';

```
